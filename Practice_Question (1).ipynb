```python
#Write a function check_even(num) that returns "Even" if the number given by user is even, otherwise "Odd".
def check_even(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
print(check_even(10))
print(check_even(7))
```

```python
#Write a function double_list(numbers) that takes a list of numbers and returns a new list with each number doubled.
#Example: [1, 2, 3] → [2, 4, 6]
def double_list(numbers):
    return [num * 2 for num in numbers]

example = [45, 55, 12]
result = double_list(example)
print(result)
```

```python
#Write a function remove_negatives(numbers) that returns a list with all negative numbers removed.
#Example: [5, -2, 7, -1] → [5, 7]
def remove_negatives(numbers):
    return [num for num in numbers if num >= 0]
example = [0, 45, -57, -91]
print(remove_negatives(example))
```

```python
#Write a function filter_even(numbers) that returns a list containing only even numbers.
#Example: [1, 2, 3, 4, 5] → [2, 4]
def filter_even(numbers):
    result = []
    for num in numbers:
        if num % 2 == 0:
            result.append(num)
    return result
example = [8,9,10,11,12,13]
print(filter_even(example))
```

```python
#Write a function word_lengths(words) that takes a list of strings and returns a list of their lengths.
#Example: ["apple", "banana", "kiwi"] → [5, 6, 4]
def word_lengths(words):
  lengths = []
  for word in words:
    lengths.append(len(word))
    return lengths
example = ["Sunflower", "Tulip", "Lily"]
print(word_lengths(example))
```

```python
#Write a function reverse_list(items) that returns the list in reverse order.
#Example: [1, 2, 3] → [3, 2, 1]
def reverse_list(items):
  return items[::-1]
example = [9, 8, 7]
print(reverse_list(example))
```

```python
#Write a function min_max(numbers) that returns both the smallest and largest number in the list.
def min_max(numbers):
  smallest = min(numbers)
  largest = max(numbers)
  return (smallest, largest)
example = [5, 2, 9, 1, 7]
print(min_max(example))
```

```python
#Write a function capitalize_words(words) that returns a new list with each word capitalized.
#Example: ["hello", "world"] → ["Hello", "World"]
def capitalize_words(words):
  result = []
  for word in words:
    result.append(word.capitalize())
    return result
example = ["gayathri", "Mahto"]
print(capitalize_words(example))
```

```python
#Write a function power(base, exponent=2) that returns base raised to the power of exponent.
#Call it with power(5) → should give 25.
#Call it with power(2, 3) → should give 8.
def power(base, exponent=2):
  return base ** exponent
print(power(5))
print(power(2, 3))
```

```python
#Write a function add_numbers(*args) that adds all numbers passed to it.
#Example: add_numbers(1, 2, 3, 4) → should print 10.
#Example: add_numbers(10, 20) → should print 30.
def add_numbers(*args):
  total = sum(args)
  print(total)
add_numbers(1, 2, 3, 4)   # Output: 10
add_numbers(10, 20)
```

```python
#Write a function print_info(**kwargs) that prints all key-value pairs.
'''	Example:
#print_info(name="Priya", age=22, city="Delhi")
Output should be:
name = Priya
age = 22
city = Delhi'''
def print_info(**kwargs):
  for key, value in kwargs.items():
        print(f"{key} = {value}")
print_info(name="Priya", age=22, city="Delhi")
```
