```python
#Write a function power_values(n) that returns both the square and cube of a number given by user.
#Example: power_values(3) → (9, 27)
def power_values(a):
  square=a*a
  cube=a*a*a
  return square,cube
result=power_values(5)
print('square: ', result[0])
print('cube: ', result[1])
```
